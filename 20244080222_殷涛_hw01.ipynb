## 欧几里得算法
实现最大公约数（GCD）的计算。

```python
# 1. 实现 gcd 函数
def gcd(m, n):
    """使用欧几里得算法计算两个整数的最大公约数。"""
    while n != 0:
        m, n = n, m % n
    return m

# 2. 测试几组数字
print("gcd(2018, 2019) =", gcd(2018, 2019))
print("gcd(1200, 300) =", gcd(1200, 300))
print("gcd(5040, 60) =", gcd(5040, 60))

# 3. 负数参数的行为
print("\n负数参数测试：")
print("gcd(-12, 18) =", gcd(-12, 18))
print("gcd(12, -18) =", gcd(12, -18))
print("gcd(-12, -18) =", gcd(-12, -18))
print("""
负数行为说明：算法会返回绝对值的GCD（可能为正或负）。\n这种行为不完全有意义，因为GCD的定义应该是正数。\n改进方法是在返回前取绝对值：return abs(m)。
""")
```

## 近似欧拉数 e
使用不同方法近似自然对数的底数 e。

```python
import math

# 1. euler_limit 函数：使用极限公式 (1+1/n)^n
def euler_limit(n):
    """使用极限公式 (1+1/n)^n 近似 e。"""
    return (1 + 1/n) ** n

print("=== euler_limit 测试 ===")
for n in [1, 10, 100, 1000, 10000]:
    print(f"n={n:5d}, euler_limit(n)={euler_limit(n):.10f}, 真实e={math.e:.10f}")

# 2. euler_infinite_sum 函数：使用无穷级数前 n 项和
def euler_infinite_sum(n):
    """使用级数 sum_{k=0}^{n-1} 1/k! 近似 e。"""
    if n == 0:
        return 0.0
    total = 0.0
    factorial = 1
    for k in range(n):
        if k > 0:
            factorial *= k
        total += 1.0 / factorial
    return total

print("\n=== euler_infinite_sum 测试 ===")
for n in range(0, 10):
    print(f"n={n}, euler_infinite_sum(n)={euler_infinite_sum(n):.10f}")

# 3. euler_approx 函数：直到误差小于 epsilon
def euler_approx(epsilon):
    """使用级数求和近似 e，直到误差小于 epsilon。"""
    true_e = math.e
    total = 0.0
    factorial = 1
    k = 0
    while True:
        if k > 0:
            factorial *= k
        term = 1.0 / factorial
        total += term
        if abs(total - true_e) < epsilon:
            break
        k += 1
    return total

print("\n=== euler_approx 测试 ===")
for eps in [0.1, 0.01, 0.001, 1e-6, 1e-10]:
    approx = euler_approx(eps)
    print(f"epsilon={eps:10f}, 近似值={approx:.10f}, 误差={abs(approx - math.e):.2e}")

# 4. print_euler_sum_table 函数：打印从 1 到 n 的级数和
def print_euler_sum_table(n):
    """打印 k 从 1 到 n 的 euler_infinite_sum(k) 值。"""
    for k in range(1, n + 1):
        print(f"k={k:2d}, sum={euler_infinite_sum(k):.10f}")

print("\n=== print_euler_sum_table(8) ===")
print_euler_sum_table(8)
```

## 测试整数属性
判断一个数是否是另一个数的幂。

```python
# 1. is_power_of_2 函数：判断 n 是否是 2 的幂
def is_power_of_2(n):
    """判断正整数 n 是否是 2 的幂。"""
    if n <= 0:
        return False
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True

print("=== is_power_of_2 测试 ===")
for n in [1, 2, 3, 4, 5, 8, 16, 18, 32, 64, 100]:
    print(f"{n:3d} 是2的幂吗？ {is_power_of_2(n)}")

# 2. is_power 函数：判断 n 是否是 b 的幂
def is_power(b, n):
    """判断正整数 n 是否是正整数 b 的幂。"""
    if b <= 1 or n <= 0:
        return False
    while n > 1:
        if n % b != 0:
            return False
        n = n // b
    return True

print("\n=== is_power 测试 ===")
test_cases = [
    (2, 1), (2, 2), (2, 8), (2, 10),
    (3, 9), (3, 27), (3, 10),
    (5, 25), (5, 125), (5, 100),
    (10, 1000), (10, 1001),
]
for b, n in test_cases:
    print(f"is_power({b}, {n}) = {is_power(b, n)}")
```
