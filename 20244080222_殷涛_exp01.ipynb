## 实验 2：函数、条件、循环与递归
函数先验证输入边界；递归和迭代版本的阶乘应得到一致结果。


```python
def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError("成绩必须在 0 到 100 之间")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"

print({score: score_level(score) for score in (95, 82, 60, 59)})
try:
    score_level(101)
except ValueError as exc:
    print("边界检查：", exc)

for_sum = sum(range(1, 101))
i, while_sum = 1, 0
while i <= 100:
    while_sum += i
    i += 1
print(f"for 求和={for_sum}, while 求和={while_sum}, 结果一致={for_sum == while_sum}")

def factorial_recursive(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("n 必须为非负整数")
    return 1 if n <= 1 else n * factorial_recursive(n - 1)

def factorial_iterative(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("n 必须为非负整数")
    result = 1
    for number in range(2, n + 1):
        result *= number
    return result

for n in (0, 1, 5, 10):
    print(f"{n}! = {factorial_recursive(n)} / {factorial_iterative(n)}")

def summary(values):
    if not values:
        raise ValueError("values 不能为空")
    return min(values), max(values), sum(values) / len(values)

minimum, maximum, average = summary([85, 58, 92, 76, 60, 89])
print(f"元组解包：最小={minimum}, 最大={maximum}, 平均={average:.2f}")

def sum_all(*args):
    return sum(args)

print("sum_all 测试：", sum_all(), sum_all(7), sum_all(1, 2, 3, 4))
```

## 实验 4：文件读写、异常处理与模块
先创建相对数据目录和 UTF-8 文件，逐行解析时保留行号和错误原因；然后生成统计报告、pickle 文件以及可导入的解析模块。


```python
import os
import pickle

# 1. 使用 os.getcwd() 查看工作目录，os.path.join() 构造相对路径，os.makedirs() 创建数据目录
print("=== 工作目录与路径 ===")
cwd = os.getcwd()
print(f"当前工作目录：{cwd}")

data_dir = os.path.join(cwd, "exp01_data")
os.makedirs(data_dir, exist_ok=True)
print(f"数据目录：{data_dir}")
print(f"目录已创建（exist_ok=True 不报错）：{os.path.exists(data_dir)}")

# 2. 使用 with open(..., encoding="utf-8") 写入数据，再读取文件，跳过表头
students_path = os.path.join(data_dir, "students.txt")
student_content = "name,score\nAlice,85\nBob,58\nCarol,92\nDavid,76\nEva,60\nFrank,89\nGrace,abc\nHenry,105\nIncomplete"

with open(students_path, "w", encoding="utf-8") as f:
    f.write(student_content)
print(f"\n已写入文件：{students_path}")

# 读取并跳过表头
print("\n=== 文件读取 ===")
with open(students_path, "r", encoding="utf-8") as f:
    lines = f.readlines()
print(f"共读取 {len(lines)} 行（含表头）")

# 3. parse_student(line) 函数
def parse_student(line):
    '将一行文本解析为学生字典，无效记录抛出 ValueError。'
    line = line.strip()
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量不对（期望2个，实际{len(parts)}个）")
    name, score_str = parts
    name = name.strip()
    score_str = score_str.strip()
    if not name:
        raise ValueError("姓名为空")
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩无法转换为整数：{score_str!r}")
    if score < 0 or score > 100:
        raise ValueError(f"成绩超出范围（0~100）：{score}")
    return {"name": name, "score": score}

# 4. 按行处理文件，捕获无效记录的异常，记录行号与错误原因，并继续处理后续记录
print("\n=== 逐行解析 ===")
valid_records = []
invalid_records = []

for line_num, line in enumerate(lines, start=1):
    if line_num == 1:
        continue  # 跳过表头
    try:
        record = parse_student(line)
        valid_records.append(record)
    except ValueError as e:
        invalid_records.append({
            "line": line_num,
            "content": line.strip(),
            "reason": str(e)
        })

print(f"有效记录数：{len(valid_records)}")
for r in valid_records:
    print(f"  {r}")
print(f"无效记录数：{len(invalid_records)}")
for r in invalid_records:
    print(f"  第{r['line']}行 [{r['content']!r}]：{r['reason']}")

# 5. 将有效记录数、无效记录数、平均分和及格率写入 report.txt，分别演示 w 与 a 模式
print("\n=== 生成报告 ===")
report_path = os.path.join(data_dir, "report.txt")
valid_count = len(valid_records)
invalid_count = len(invalid_records)
avg_score = sum(r["score"] for r in valid_records) / valid_count
pass_rate_val = sum(1 for r in valid_records if r["score"] >= 60) / valid_count

# w 模式：覆盖写入
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数：{valid_count}\n")
    f.write(f"无效记录数：{invalid_count}\n")
    f.write(f"平均分：{avg_score:.2f}\n")
    f.write(f"及格率：{pass_rate_val:.2%}\n")
print(f"w 模式写入完成：{report_path}")

# a 模式：追加写入
with open(report_path, "a", encoding="utf-8") as f:
    f.write("---\n")
    f.write("说明：w 模式会覆盖文件原有内容；a 模式在文件末尾追加内容，不覆盖已有数据。\n")
print("a 模式追加完成")

with open(report_path, "r", encoding="utf-8") as f:
    print("\n报告内容：")
    print(f.read())

# 6. 使用 pickle.dump() 保存有效记录，pickle.load() 读取并验证
print("=== pickle 序列化 ===")
pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)
print(f"已保存：{pkl_path}")

with open(pkl_path, "rb") as f:
    restored = pickle.load(f)
print(f"读取恢复数据，与原数据相等：{restored == valid_records}")

# 7. 从 student_utils.py 模块导入 parse_student，并测试有效和无效输入
print("\n=== 模块导入测试 ===")
# 先确保 student_utils.py 存在
import importlib
if "student_utils" in dir():
    importlib.reload(student_utils)
else:
    import student_utils

print("有效输入测试：student_utils.parse_student('Iris,88') =", student_utils.parse_student('Iris,88'))
try:
    student_utils.parse_student("Iris,wrong")
except ValueError as e:
    print(f"无效输入测试：抛出 ValueError：{e}")

# 8. 读取一个明确不存在的文件，捕获 FileNotFoundError
print("\n=== 文件不存在异常 ===")
missing_path = os.path.join(data_dir, "not_exist.txt")
try:
    with open(missing_path, "r", encoding="utf-8") as f:
        f.read()
except FileNotFoundError as e:
    print(f"捕获 FileNotFoundError：{e}")
    print("说明：尝试打开不存在的文件时会触发此异常，应在代码中捕获并给出友好提示。")
```

## 实验 5：类、继承与对象复制
类封装学生状态；子类以 `super()` 复用父类初始化。对含嵌套列表的对象分别采用直接赋值、浅拷贝和深拷贝来观察引用关系。


```python
import copy

class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score}, passed={self.is_passed()})"

class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return f"GraduateStudent(name={self.name}, score={self.score}, topic={self.research_topic}, passed={self.is_passed()})"

students = [Student(item["name"], item["score"]) for item in valid_records]
for student in students:
    print(student)
graduate = GraduateStudent("Iris", 91, "数据分析")
print(graduate)

def original_object():
    return {"name": "Alice", "scores": [85, 90]}

direct = original_object()
direct_alias = direct
direct_alias["scores"].append(100)
print("直接赋值：", direct)

shallow = original_object()
shallow_copy = copy.copy(shallow)
shallow_copy["scores"].append(100)
print("浅拷贝：原对象=", shallow, "浅拷贝=", shallow_copy)

deep = original_object()
deep_copy = copy.deepcopy(deep)
deep_copy["scores"].append(100)
print("深拷贝：原对象=", deep, "深拷贝=", deep_copy)
```
